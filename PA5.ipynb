# Building Facilities and Sets 

```python
b1facilities = set()
while True:
    facility = input('Enter a facility for the first building (“Done” to stop): ')
    if facility.lower() == "done":
        break
    b1facilities.add(facility)
```

```python
b2facilities = set()
while True:
    facility = input('Enter a facility for the second building (“Done” to stop): ')
    if facility.lower() == "done":
        break
    b2facilities.add(facility)
```

```python
print("\nFacilities in the first building:", b1facilities)
print("Facilities in the second building:", b2facilities)
```

```python
print("\nFacilities in both buildings:", b1facilities & b2facilities)
```

```python
print("Facilities in at least one building:", b1facilities | b2facilities)
```

# Triangle Calculations 

```python
def triangle_solve(leg1, leg2):
    hypotenuse = (leg1**2 + leg2**2) ** 0.5
    perimeter = leg1 + leg2 + hypotenuse
    area = 0.5 * leg1 * leg2

    return (round(hypotenuse), round(perimeter), round(area))
```

```python
triangle_solve(4,5)
```

# Calculator Class 

```python
class Calculator:
    def __init__(self):
        self.result_recent = None
        self.total_calculations = 0

    def add(self, a, b):
        self.result_recent = a + b
        self.total_calculations += 1
        return self.result_recent

    def subtract(self, a, b):
        self.result_recent = a - b
        self.total_calculations += 1
        return self.result_recent

    def multiply(self, a, b):
        self.result_recent = a * b
        self.total_calculations += 1
        return self.result_recent

    def divide(self, a, b):
        if b == 0:
            raise ValueError("Cannot divide by zero.")
        self.result_recent = a / b
        self.total_calculations += 1
        return self.result_recent

    def floor_divide(self, a, b):
        if b == 0:
            raise ValueError("Cannot divide by zero.")
        self.result_recent = a // b
        self.total_calculations += 1
        return self.result_recent

    def status(self):
        print(f"Most Recent Result: {self.result_recent}")
        print(f"Total Calculations: {self.total_calculations}")
```

```python
calc = Calculator()
print(calc.add(10, 5))
print(calc.subtract(10, 5))
print(calc.multiply(3, 4))
print(calc.divide(8, 2))
print(calc.floor_divide(-7, 3))
calc.status()
```
