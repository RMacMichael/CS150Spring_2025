#Task 1: Input Positive Integer and Factors 

```python
num = int(input("input a positive integer:"))
counter = 0
for i in range(1, num + 1):
    if num % i == 0:
        print(f"{i} is a factor of {num}")
        counter += 1
if counter == 2:  
    print(f"{num} is prime")
else:
    print(f"{num} is not prime")
```

#Task 2: Language Translator 

```python
translation_dict = {
    "chair": "silla",
    "table": "mesa",
    "dog": "perro",
    "house": "casa",
    "car": "coche",
    "book": "libro",
    "water": "agua",
    "sun": "sol",
    "moon": "luna",
    "sorrow": "pena"}
while True:
    word = input("Enter a word to translate from English to Spanish (hit enter to stop): ").lower()
    if word == "":
        break
    if word in translation_dict:
        print(f"The Spanish word for {word} is {translation_dict[word].upper()}.")
    else:
        print(f"The Spanish word for {word} is not in this dictionary.")
```

```python
employee_dict = {}
print("Enter 7 employee names and their corresponding positions.")
for _ in range(8):
    name = input("Enter employee name: ")
    position = input(f"Enter position for {name}: ")
    employee_dict[name] = position
while True:
    print("Menu Options:")
    print("1. Find an employee’s title")
    print("2. Find all employees with a given title")
    print("3. Quit the program")

    choice = input("Enter your choice (1-3): ")
    if choice == "1":
        search_name = input("Enter the employee's name: ")
        if search_name in employee_dict:
            print(f"{search_name} is a {employee_dict[search_name]}.")
        else:
            print("No one by that name is present.")
    elif choice == "2":
        search_position = input("Enter a job title: ")
        matching_employees = [name for name, title in employee_dict.items() if title == search_position]  
        if matching_employees:
            print(f"Employees with the title '{search_position}': {', '.join(matching_employees)}")
        else:
            print("No employees hold this position.")
    elif choice == "3":
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 3.")
```
