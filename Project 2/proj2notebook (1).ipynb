```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("C:\\Users\\12486\\Downloads\\KO.csv")

col1 = 'Open'
col2 = 'Close'

def compute_stats(series):
    stats = {
        'Mean': series.mean(),
        'Max': series.max(),
        'Min': series.min(),
        'Standard Deviation': series.std(),
        'Range': series.max() - series.min(),
        'Median': series.median()
    }
    return stats

print(f"Statistics for {col1}:")
print(compute_stats(df[col1]))
print()
print(f"Statistics for {col2}:")
print(compute_stats(df[col2]))
print()

df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month_name()

for month in df['Month'].unique():
    print(f"Stats for {month}:")
    
    month_data = df[df['Month'] == month]
    open_mean = month_data['Open'].mean()
    open_max = month_data['Open'].max()
    open_min = month_data['Open'].min()
    open_std = month_data['Open'].std()
    
    close_mean = month_data['Close'].mean()
    close_max = month_data['Close'].max()
    close_min = month_data['Close'].min()
    close_std = month_data['Close'].std()

    print(f"  Open Mean: {open_mean}")
    print(f"  Open Max: {open_max}")
    print(f"  Open Min: {open_min}")
    print(f"  Open Std: {open_std}")
    print()
    print(f"  Close Mean: {close_mean}")
    print(f"  Close Max: {close_max}")
    print(f"  Close Min: {close_min}")
    print(f"  Close Std: {close_std}")
    print() 

plt.scatter(df['Open'], df['Close'])
plt.xlabel('Open Price')  
plt.ylabel('Close Price')  
plt.title('Scatter Plot of Open vs Close Prices')
plt.show()

plt.hist(df['Open'], color='cyan')
plt.title('Histogram of Open Prices')
plt.xlabel('Open Price')
plt.ylabel('Frequency')
plt.show()


plt.hist(df['Close'], color='purple')
plt.title('Histogram of Close Prices')
plt.xlabel('Close Price')
plt.ylabel('Frequency')
plt.show()

df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
years = sorted(df['Year'].unique())[:2]

for year in years:
    year_data = df[df['Year'] == year]
    plt.hist(year_data['Open'], color='green')
    plt.title(f'Open Prices in {year}')
    plt.xlabel('Open Price')
    plt.ylabel('Frequency')
    plt.show()
```
